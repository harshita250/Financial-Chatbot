```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("financials.csv")

st.title("📊 Financial Data Chatbot")
st.write("Ask questions like 'Show Apple revenue in 2023' or 'Plot MSFT net income'")

user_input = st.text_input("Enter your question:")

def parse_query(query):
    query = query.lower()
    if "plot" in query:
        return "plot"
    elif "show" in query or "get" in query:
        return "show"
    elif "export" in query or "pdf" in query:
        return "export"
    return "unknown"

def extract_entities(query):
    query = query.lower()
    company = None
    if "apple" in query or "aapl" in query:
        company = "AAPL"
    elif "msft" in query or "microsoft" in query:
        company = "MSFT"

    year = next((y for y in [2022, 2023, 2024] if str(y) in query), None)

    column = None
    for col in df.columns:
        if col.lower() in query:
            column = col
            break

    return company, year, column

def show_data(company, year, column):
    result = df.copy()

    if company:
        result = result[result['Company'].str.upper() == company.upper()]

    if year:
        result = result[result['Year'] == int(year)]

    if column:
        if column in result.columns:
            if result.empty:
                return f"No data found for {company} in {year}."
            return result[['Company', 'Year', column]]
        else:
            return f"Column '{column}' not found."

    return result if not result.empty else "No matching data found."

def plot_data(company, column):
    data = df[df['Company'].str.upper() == company.upper()]
    if column not in data.columns:
        st.warning(f"Column '{column}' not found for plotting.")
        return
    fig, ax = plt.subplots()
    ax.plot(data['Year'], data[column], marker='o')
    ax.set_title(f"{company} - {column} over time")
    ax.set_xlabel("Year")
    ax.set_ylabel(column)
    st.pyplot(fig)

if user_input:
    intent = parse_query(user_input)
    company, year, column = extract_entities(user_input)

    if intent == "show":
        result = show_data(company, year, column)
        st.write(result)

    elif intent == "plot":
        if company and column:
            plot_data(company, column)
        else:
            st.warning("Please specify both company and metric to plot.")

    elif intent == "export":
        st.info("PDF export feature is coming soon!")

    else:
        st.warning("Sorry, I didn't understand your request. Try asking to 'show' or 'plot' data.")
```
