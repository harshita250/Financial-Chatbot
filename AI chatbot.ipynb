```python
import pandas as pd
```

```python
df = pd.read_csv('BCGX.csv')
df.head()
```

```python
df.head()
```

```python
df = df.sort_values(by=["Company", "Year"], ascending=[True, True]).reset_index(drop=True)
```

```python
df.info()
```

```python
# Function to convert M/B strings to billions
def convert_to_billions(x):
    if isinstance(x, str) and x.upper().endswith('M'):
        return float(x[:-1]) / 1000  # millions to billions

# Apply to all financial columns
cols_to_convert = ["Total Revenue", "Net Income", "Total Assets", "Total Liabilities", "Net Cash Flow from Operating"]
df[cols_to_convert] = df[cols_to_convert].applymap(convert_to_billions)

# Round for clarity
df[cols_to_convert] = df[cols_to_convert].round(2)
```

```python
df
```

```python
#YoY Trends
metrics = ["Total Revenue", "Net Income", "Total Assets", "Total Liabilities", "Net Cash Flow from Operating"]

for col in metrics:
    df[f"{col} YoY %"] = df.groupby("Company")[col].pct_change() * 100  # multiply by 100 for %
    df[f"{col} YoY %"] = df[f"{col} YoY %"].round(2)  # optional rounding
```

```python
df.fillna(0, inplace=True)
```

```python
df
```

### 📊 Apple Financial Performance Summary (2022–2024)

- 2023 Dip:
Apple saw a decline in both revenue **(-2.8%)** and net income **(-2.8%)** in 2023. Operating cash flow dropped ~9.5%, indicating softer business performance and possible operational inefficiencies.
- 2024 Rebound (Partially):
Revenue recovered **(+2.0%)** and cash flow from operations rose by ~7%, showing signs of stabilization.
However, net income continued to decline **(-3.4%)**, suggesting profitability pressures — likely due to rising costs or margin erosion.
- Balance Sheet:
Assets grew modestly in 2024 **(+3.5%)**, while liabilities jumped 6%, indicating a possible uptick in borrowing or deferred expenses.

### 📊 Microsoft Financial Performance Summary (2022–2024)

- 2023: Stable Revenue, Flat Profits
Microsoft grew its revenue by **6.9%**, but net income slightly declined **(-0.5%)**, suggesting rising costs or flat margins. Cash flow from operations also dipped 1.6%, a minor operational signal.
- 2024: Strong Growth Across the Board
Revenue surged **15.7%**, and net income jumped **21.8%**, showing not just topline expansion but also improved profitability.
Operating cash flow soared **35.4%**, signaling efficient and profitable core operations.
- Balance Sheet Strength:
Assets grew **24.3%** in 2024, matched by an **18.4%** rise in liabilities.
Still, liabilities are less than 50% of revenue, indicating a healthy leverage position and good solvency.

### 📊Tesla Financial Performance Summary (2022–2024)

- 2023: Strong Revenue, Higher Debt
Revenue jumped **18.8%**, and net income grew **18.9%**, signaling good momentum.
However, assets shrank **(-15.3%)** while liabilities surged **(+70.3%)**, indicating Tesla may have taken on substantial new debt or restructured its balance sheet.
Operating cash flow dipped **(-9.9%)**, showing some stress in core operations.
- 2024: Profit Collapse, Slight Operational Recovery
Revenue stayed flat **(+0.95%)**, but net income fell sharply (-52.2%) — likely due to increased costs or lower margins.
Cash flow from operations improved **+12.5%**, hinting at internal efficiency gains, despite the profit drop.
Assets continued to decline **(-21%)**, while liabilities increased again **(+7.4%)**.
